# Loop for all Fingerprint Combinations (sets of 3)

The following code creates all possible combinations of 3 fingerprint categories with weights between 0 and 12 (inclusive), with as much elimination of repeats as possible (i.e. A-2,B-4,C-4 would be equivalent to A-1,B-2,C-2).
