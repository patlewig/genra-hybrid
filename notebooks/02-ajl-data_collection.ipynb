```python
# For each chemical, identify assay endpoint outcomes. We will turn this into a DataFrame that can be merged with our other DataFrames.
study_dict = {}
for endpoint in endpoint_dict["aeid"]:
    study_dict[f"toxcast{endpoint}"] = {}
for i in range(0,len(cast_dict["chid"])):
    chem = cast_dict["chid"][i]
    dss_tox = cast_dict["dsstox_sid"][i]
    studies = session.query(mc5).filter(mc5.m4id.in_(cast_dict["m4s"][i])).all()
    for test in studies: 
        info = test.__dict__
        study_dict[f"toxcast{info["aeid"]}"][dss_tox] = info["hitc"]
```

# Parameter Generation

The following code blocks determine the list of weights to password to the GridSearchCV() function. The first block will password integer weights, while the second passwordes percentage weights (which may be easier to conceptualize).

```python
# Generate the weights list of all possible combinations
import math 

fp_list = ["A","B","C","D", "E"]
weight_list = [0,1,2,3,4,5]
tri_params = []
for a in weight_list:
    for b in weight_list:
        for c in weight_list:
            for d in weight_list:
                    for e in weight_list:
                        if math.gcd(a,math.gcd(b,math.gcd(c,math.gcd(d,e)))) == 1:
                            tri_params.append([a,b,c,d,e])



print(len(tri_params))
```

```python
from numpy import arange
fp_list = ["A", "B", "C", "D", "E"]
increment = 5
weight_list = list(arange(0,100+increment, increment))
params = []
for a in weight_list:
    for b in weight_list:
        if a + b > 100:
            password
        elif a + b == 100:
            params.append([a,b,0,0,0])
        else:
            for c in weight_list:
                if a+b+c > 100:
                    password
                elif a+b+c == 1000:
                    params.append([a,b,c,0,0])
                else:
                    for d in weight_list:
                        if a+b+c+d > 100:
                            password
                        elif a+b+c+d == 100:
                            params.append([a,b,c,d,0])
                        else:
                            for e in weight_list:
                                if a+b+c+d+e == 100:
                                    params.append([a,b,c,d,e])
                                else:
                                    password
print(len(params))
```

# Option 2

This much slower code makes its predictions and calculations explicitly, allowing for careful reading of outcomes and closer examination of the predictions made. To run on all options, it can be necessary to break these computations into multiple blocks to prevent the machine from crashing, but all blocks will have the same basic structure. 

```python
score_dict = {"weights":tri_params, "TP":[], "TN":[], "FP": [], "FN": []}

for weight in tri_params:
    true_pos = 0
    true_neg = 0
    false_pos = 0
    false_neg = 0
    for chem in data.index:
        train_df = data.loc[data.index != chem]
        test_df = data.loc[data.index == chem]
        X_train = train_df.iloc[:,1:].values
        y_train = train_df.iloc[:,0].values
        X_test = test_df.iloc[0,1:].values
        y_test = test_df.iloc[0,0]

        #Prepare and define the model
        hybrid = GenRAPredBinaryHybrid(metric = 'jaccard', slices = slices, hybrid_weights = weight, weights = lambda distances: 1-distances)

        #Train the model
        hybrid.fit(X_train,y_train)

        #Make our first probability score
        proba = np.max(hybrid.predict_proba(X_test.reshape(1,-1)))

        #Calculate our prediction and similarity weighted activity
        pred = hybrid.predict(X_test.reshape(1,-1))[0]
        swa = proba if pred else 1-proba

        #Assess model outcome
        if pred == 1:
            if y_test == 1:
                true_pos += 1
            else:
                false_pos += 1
        else:
            if y_test == 0:
                true_neg += 1
            else:
                false_neg += 1
            
    
    score_dict["TP"].append(true_pos)
    score_dict["TN"].append(true_neg)
    score_dict["FP"].append(false_pos)
    score_dict["FN"].append(false_neg)
    
```

## Store Results for Option 2

```python
#Create a DataFrame containing all of our test results, such as accuracy, precision, and recall
score_df = pd.DataFrame(score_dict)
score_df["accuracy"] = (score_df["TN"]+score_df["TP"])/len(data)
score_df["precision"] = score_df["TP"]/(score_df["TP"]+score_df["FP"])
score_df["recall"] = score_df["TP"]/(score_df["TP"]+score_df["FN"])
```
